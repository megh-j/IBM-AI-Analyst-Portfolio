# 1. K-Means Clustering

**Problem Statement:** Create a dataset containing Age and Spending Score of 8 customers.
Apply K-Means Clustering with K = 2 and display the cluster labels.

```python
import pandas as pd
from sklearn.cluster import KMeans

data = {
    'Age':            [22, 25, 24, 28, 60, 65, 58, 62],
    'Spending_Score': [90, 85, 88, 78, 15, 20, 10, 25]
}

df = pd.DataFrame(data)
X = df[['Age', 'Spending_Score']]

model = KMeans(n_clusters=2, random_state=42)
cluster_labels = model.fit_predict(X)

df['Cluster'] = cluster_labels
print("Customer Dataset with K-Means Cluster Labels:")
print("-" * 45)
print(df)
```

# Hierarchical Clustering

**Problem Statement:** Create a dataset with Age and Income values for 8 people. Apply
Agglomerative Clustering with 2 clusters and display the cluster assignments

```python
import pandas as pd
from sklearn.cluster import AgglomerativeClustering

data = {
    'Age':    [22, 24, 25, 27, 45, 48, 50, 55],
    'Income': [25, 30, 28, 35, 80, 85, 90, 95]
}

df = pd.DataFrame(data)
X = df[['Age', 'Income']]
model = AgglomerativeClustering(n_clusters=2)

cluster_labels = model.fit_predict(X)
df['Cluster'] = cluster_labels

print("Hierarchical Clustering Assignments:")
print("-" * 40)
print(df)
```

# Dendrogram Visualization

**Problem Statement:** Generate a Dendrogram
using the Ward linkage method

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

data = {
    'Age':    [22, 24, 25, 27, 45, 48, 50, 55],
    'Income': [25, 30, 28, 35, 80, 85, 90, 95]
}
df = pd.DataFrame(data)

Z = linkage(df, method='ward')
plt.figure(figsize=(10, 6))
plt.title("Hierarchical Clustering Dendrogram (Ward Linkage)")
plt.xlabel("Customer Index (Row Number)")
plt.ylabel("Distance")

dendrogram(Z, labels=df.index)

plt.show()
```

# DBSCAN Clustering

**Problem Statement:**  Create the following dataset: X = [1,1,2,2,8,8,9,9,50,55] Y =
[1,2,1,2,1,2,1,2,100,100] Apply DBSCAN with: • eps = 3 • min_samples = 2 Display the cluster
labels and identify any outliers.

```python
import pandas as pd
from sklearn.cluster import DBSCAN

data = {
    'X': [1, 1, 2, 2, 8, 8, 9, 9, 50, 55],
    'Y': [1, 2, 1, 2, 1, 2, 1, 2, 100, 100]
}

df = pd.DataFrame(data)
model = DBSCAN(eps=3, min_samples=2)

df['Cluster'] = model.fit_predict(df[['X', 'Y']])
print("DBSCAN Clustering Results:")
print("-" * 30)
print(df)
print("\n")

outliers = df[df['Cluster'] == -1]
print("Identified Outliers:")
print("-" * 30)
print(outliers)
```

Create your own dataset with: • Age • Income Apply K-Means, Hierarchical Clustering, and
DBSCAN on the same dataset and compare the results.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN

data = {
    'Age':    [22, 23, 24, 25, 26,  45, 46, 48, 49, 50,  60, 62, 64, 65, 68,  25, 65],
    'Income': [20, 28, 25, 22, 24,  85, 92, 90, 86, 88,  50, 55, 52, 45, 48,  95, 15]
}

df = pd.DataFrame(data)
X = df[['Age', 'Income']]

kmeans = KMeans(n_clusters=3, random_state=42)
df['KMeans_Labels'] = kmeans.fit_predict(X)

hierarchical = AgglomerativeClustering(n_clusters=3)
df['Hierarchical_Labels'] = hierarchical.fit_predict(X)

dbscan = DBSCAN(eps=15, min_samples=3)
df['DBSCAN_Labels'] = dbscan.fit_predict(X)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle('Clustering Algorithm Comparison', fontsize=16)

axes[0].scatter(df['Age'], df['Income'], c=df['KMeans_Labels'], cmap='viridis', s=100)
axes[0].set_title('K-Means Clustering')
axes[0].set_xlabel('Age')
axes[0].set_ylabel('Income ($1000s)')

axes[1].scatter(df['Age'], df['Income'], c=df['Hierarchical_Labels'], cmap='plasma', s=100)
axes[1].set_title('Hierarchical Clustering')
axes[1].set_xlabel('Age')

axes[2].scatter(df['Age'], df['Income'], c=df['DBSCAN_Labels'], cmap='coolwarm', s=100)
axes[2].set_title('DBSCAN (Outliers in distinct color)')
axes[2].set_xlabel('Age')

plt.show()
print(df)
```
