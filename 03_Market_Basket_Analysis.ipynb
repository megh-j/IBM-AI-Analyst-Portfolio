# Apriori

**Problem Statement:** Create your own dataset with at least 10
supermarket transactions containing items such as Milk, Bread, Eggs, Butter, Juice, etc. Using
the Apriori algorithm: Find frequent itemsets with min_support = 0.3. Generate association
rules. Identify the rule with the highest confidence.

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

dataset = [
    ['Milk', 'Bread', 'Eggs', 'Butter'],
    ['Milk', 'Bread', 'Eggs'],
    ['Bread', 'Butter', 'Juice'],
    ['Milk', 'Bread', 'Butter', 'Juice'],
    ['Milk', 'Eggs', 'Juice'],
    ['Bread', 'Eggs', 'Butter'],
    ['Milk', 'Bread', 'Eggs', 'Butter', 'Juice'],
    ['Milk', 'Bread', 'Butter'],
    ['Eggs', 'Juice'],
    ['Milk', 'Bread', 'Eggs', 'Juice']
]

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

print("--- TRANSFORMED DATASET (First 5 Rows) ---")
print(df.head(), "\n")

frequent_itemsets = apriori(df, min_support=0.3, use_colnames=True)

print("--- FREQUENT ITEMSETS (Support >= 0.3) ---")
print(frequent_itemsets.sort_values(by='support', ascending=False).to_string(index=False))
print("\n" + "="*50 + "\n")

rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)

print("--- ASSOCIATION RULES (Confidence & Lift) ---")

columns_to_show = ['antecedents', 'consequents', 'support', 'confidence', 'lift']
print(rules[columns_to_show].sort_values(by='confidence', ascending=False).to_string(index=False))
print("\n" + "="*50 + "\n")

highest_rule = rules.sort_values(by='confidence', ascending=False).iloc[0]

print("--- RULE WITH THE HIGHEST CONFIDENCE ---")

antecedent = ", ".join(list(highest_rule['antecedents']))
consequent = ", ".join(list(highest_rule['consequents']))

print(f"If a customer buys [{antecedent}], they are highly likely to buy [{consequent}].")
print(f"Confidence: {highest_rule['confidence'] * 100:.1f}%")
print(f"Lift: {highest_rule['lift']:.2f}")
```

# FP Growth

**Problem Statement:** Create a dataset of 15 cafeteria
transactions made by students. Example items: Sandwich Tea Coffee Burger Juice Chips Using
FP-Growth: Find frequent itemsets. Compare the results with Apriori. Identify the most
frequently purchased item combination

```python
import pandas as pd
import time
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth, apriori

dataset = [
    ['Sandwich', 'Tea', 'Chips'],
    ['Burger', 'Juice', 'Chips'],
    ['Sandwich', 'Coffee', 'Chips'],
    ['Sandwich', 'Tea', 'Chips'],
    ['Burger', 'Coffee'],
    ['Sandwich', 'Tea', 'Juice'],
    ['Burger', 'Juice', 'Chips'],
    ['Sandwich', 'Coffee', 'Chips'],
    ['Sandwich', 'Tea'],
    ['Burger', 'Tea', 'Chips'],
    ['Sandwich', 'Coffee', 'Juice'],
    ['Burger', 'Juice'],
    ['Sandwich', 'Tea', 'Chips'],
    ['Coffee', 'Juice', 'Chips'],
    ['Sandwich', 'Tea', 'Chips']
]

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

print("--- TRANSFORMED CAFETERIA DATASET (First 5 Rows) ---")
print(df.head(), "\n")


print("--- 1. FP-GROWTH FREQUENT ITEMSETS ---")
start_fp = time.time()
fp_frequent_itemsets = fpgrowth(df, min_support=0.2, use_colnames=True)
time_fp = time.time() - start_fp

print(fp_frequent_itemsets.sort_values(by='support', ascending=False).to_string(index=False))
print(f"\n[FP-Growth Execution Time: {time_fp:.5f} seconds]\n")
print("="*50 + "\n")

print("--- 2. APRIORI FREQUENT ITEMSETS (For Comparison) ---")
start_ap = time.time()
ap_frequent_itemsets = apriori(df, min_support=0.2, use_colnames=True)
time_ap = time.time() - start_ap

print(ap_frequent_itemsets.sort_values(by='support', ascending=False).to_string(index=False))
print(f"\n[Apriori Execution Time: {time_ap:.5f} seconds]\n")
print("="*50 + "\n")

fp_frequent_itemsets['length'] = fp_frequent_itemsets['itemsets'].apply(lambda x: len(x))
combinations = fp_frequent_itemsets[fp_frequent_itemsets['length'] >= 2]

most_frequent_combo = combinations.sort_values(by='support', ascending=False).iloc[0]

print("--- 3. MOST COMMON ITEM COMBINATION ---")
combo_items = ", ".join(list(most_frequent_combo['itemsets']))
print(f"The most frequently purchased combination of items is: [{combo_items}]")
print(f"Support Value: {most_frequent_combo['support']:.3f} (Appeared in {int(most_frequent_combo['support'] * 15)} out of 15 transactions)")
```

**Problem Statement:** Create a dataset of 20 online shopping
orders. Possible products: Mobile Earphones Charger Laptop Mouse Keyboard USB Drive
Using either Apriori or FP-Growth: Find frequent itemsets. Generate association rules. Find
one rule with: Support > 0.3 Confidence > 0.7 Interpret the business meaning of the rule.
Example Interpretation "If customers buy a Laptop, they are also likely to buy a Mouse."

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

dataset = [
    ['Laptop', 'Mouse', 'USB Drive'],
    ['Laptop', 'Mouse', 'Keyboard'],
    ['Laptop', 'Mouse', 'Earphones'],
    ['Laptop', 'Mouse'],
    ['Laptop', 'Mouse', 'Keyboard', 'USB Drive'],
    ['Laptop', 'Mouse'],
    ['Laptop', 'Mouse', 'Charger'],
    ['Laptop', 'Mouse'],
    ['Laptop', 'Keyboard'],
    ['Laptop', 'USB Drive'],
    ['Mobile', 'Charger', 'Earphones'],
    ['Mobile', 'Charger', 'Earphones'],
    ['Mobile', 'Charger'],
    ['Mobile', 'Charger', 'USB Drive'],
    ['Mobile', 'Charger'],
    ['Mobile', 'Charger', 'Earphones'],
    ['Mobile', 'Charger'],
    ['Mobile', 'Earphones'],
    ['Mobile', 'USB Drive'],
    ['Mouse', 'Keyboard', 'USB Drive']
]

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

frequent_itemsets = apriori(df, min_support=0.2, use_colnames=True)

print("--- 1. FREQUENT ITEMSETS ---")
print(frequent_itemsets.sort_values(by='support', ascending=False).to_string(index=False))
print("\n" + "="*50 + "\n")

rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)

print("--- 2. ALL GENERATED RULES ---")
columns_to_show = ['antecedents', 'consequents', 'support', 'confidence', 'lift']
print(rules[columns_to_show].sort_values(by='confidence', ascending=False).to_string(index=False))
print("\n" + "="*50 + "\n")

target_rules = rules[(rules['support'] > 0.3) & (rules['confidence'] > 0.7)]

print("--- 3. RULE MATCHING CRITERIA (Support > 0.3, Confidence > 0.7) ---")
if not target_rules.empty:
  best_rule = target_rules.sort_values(by='confidence', ascending=False).iloc[0]
  antecedent = ", ".join(list(best_rule['antecedents']))
  consequent = ", ".join(list(best_rule['consequents']))

  print(f"Rule: If a customer buys a [{antecedent}], they also buy a [{consequent}].")
  print(f"Support: {best_rule['support']:.2f}")
  print(f"Confidence: {best_rule['confidence']:.2f}")

  print("\n--- 4. BUSINESS INSIGHT / INTERPRETATION ---")
  print(f"Interpretation: {best_rule['confidence']*100:.0f}% of the customers who purchased a {antecedent} "
        f"also ended up purchasing a {consequent}. Furthermore, this specific combination appeared in "
        f"{best_rule['support']*100:.0f}% of all transactions recorded.")
  print(f"Actionable Insight: The business should bundle the {consequent} with the {antecedent} at a slight "
        f"discount, or place the {consequent} prominently on the {antecedent}'s checkout page as a "
        f"1-click upsell to maximize revenue.")
else:
    print("No rules matched the specific criteria.")
```
