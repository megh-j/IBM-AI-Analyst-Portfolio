# PCA: Principal Component Analysis

**Problem Statement:** Create a dataset containing information for 20
students with the following columns: Study Hours Attendance (%) Assignments Completed
Internal Marks Final Exam Marks Apply PCA to reduce the dataset from 5 features to 2
principal components. Tasks Standardize the data. Apply PCA (n_components=2). Display the
transformed dataset

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

np.random.seed(42)

data = {
    'Study_Hours': np.random.randint(1, 10, 20),
    'Attendance_Pct': np.random.randint(50, 100, 20),
    'Assignments_Completed': np.random.randint(3, 15, 20),
    'Internal_Marks': np.random.randint(20, 50, 20),
    'Final_Exam_Marks': np.random.randint(40, 100, 20)
}

df = pd.DataFrame(data)

print("--- 1. ORIGINAL DATASET (First 5 Rows) ---")
print(df.head(), "\n")

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

pca = PCA(n_components=2)
pca_transformed = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(data=pca_transformed, columns=['Principal Component 1', 'Principal Component 2'])

print("--- 2. SHAPE COMPARISON ---")
print(f"Original shape: {df.shape} (20 rows, 5 columns)")
print(f"Reduced shape: {pca_df.shape}  (20 rows, 2 columns)\n")

print("--- 3. PCA-TRANSFORMED VALUES (All 20 Students) ---")
print(pca_df)

explained_variance = sum(pca.explained_variance_ratio_) * 100
print(f"\n[Note: These 2 components retain {explained_variance:.2f}% of the information from the original 5 features!]")
```

**Problem Statement:** Create your own employee dataset with at least 15
records. Columns: Age Salary Experience Training Hours Performance Score Apply PCA and
answer: How many features were reduced? How much variance is explained by the first two
components? Does PCA retain more than 90% of the information? Expected Output
Explained variance ratio Total variance retained Reduced dataset

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

np.random.seed(42)

age = np.random.randint(22, 60, 15)
experience = np.maximum(0, age - 22 + np.random.randint(-3, 3, 15))
performance = np.random.randint(1, 11, 15)
salary = 40000 + (experience * 3000) + (performance * 1000) + np.random.randint(-5000, 5000, 15)
training_hours = np.random.randint(10, 100, 15)

data = {
    'Age': age,
    'Salary': salary,
    'Experience': experience,
    'Training_Hours': training_hours,
    'Performance_Score': performance
}

df = pd.DataFrame(data)

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

pca = PCA(n_components=2)
pca_transformed = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(data=pca_transformed, columns=['PC1', 'PC2'])
explained_variance_ratio = pca.explained_variance_ratio_
total_variance = sum(explained_variance_ratio) * 100

print("--- 1. REDUCED DATASET (First 5 Rows) ---")
print(pca_df.head(), "\n")

print("--- 2. PCA ASSIGNMENT ANSWERS ---")
print("Q: How many features were reduced?")
print("A: 5 original features (Age, Salary, Exp, Training, Performance) were reduced to 2 principal components.\n")

print("Q: How much variance is explained by the first two components?")
print(f"A: Component 1 explains: {explained_variance_ratio[0]*100:.2f}%")
print(f"   Component 2 explains: {explained_variance_ratio[1]*100:.2f}%")
print(f"   Total Variance Retained: {total_variance:.2f}%\n")

print("Q: Does PCA retain more than 90% of the information?")
if total_variance > 90:
    print(f"A: YES. It retains {total_variance:.2f}%, which is greater than 90%.")
else:
    print(f"A: NO. It retains {total_variance:.2f}%, which is less than 90%.")
```
